```python
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('C:\\datasets\\users.csv')
```

```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop_duplicates()
df_clean = df_clean.reset_index(drop=True)
```

```python
df_clean['signup_date'] = df_clean['signup_date'].str.strip()
df_clean['signup_date'] = pd.to_datetime(df_clean['signup_date'], format='mixed', errors='coerce')
```

```python
df_clean['traffic_source'] = df_clean['traffic_source'].str.strip()
df_clean['traffic_source'] = df_clean['traffic_source'].str.upper()

df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == '-', np.nan, df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'N/A', np.nan, df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'NONE', np.nan, df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'NULL', np.nan, df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'UNKNOWN', np.nan, df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == '', np.nan, df_clean['traffic_source'])

df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'AFFILIATE', 'Affiliate', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'AFFILIATES', 'Affiliate', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'DIRECT', 'Direct', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'EMAIL', 'Email', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'E-MAIL', 'Email', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'ORGANIC', 'Organic Search', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'ORGANIC SEARCH', 'Organic Search', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'ORGANIC_SEARCH', 'Organic Search', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'PAID SEARCH', 'Paid Search', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'PAID_SEARCH', 'Paid Search', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'PAID-SEARCH', 'Paid Search', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'REFERRAL', 'Referral', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'REFERAL', 'Referral', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'SOCIAL', 'Social Media', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'SOCIAL MEDIA', 'Social Media', df_clean['traffic_source'])
df_clean['traffic_source'] = np.where(df_clean['traffic_source'] == 'SOCIAL_MEDIA', 'Social Media', df_clean['traffic_source'])
```
